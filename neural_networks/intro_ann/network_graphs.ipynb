```python
import matplotlib.pyplot as plt
import networkx as nx
from pathlib import Path
```

```python
Path("outputs").mkdir(parents=True, exist_ok=True)
```

```python
layers = {
    "Input layer":  ["x1", "x2"],
    "Layer 1\n(Hidden layer)": ["h1_1", "h1_2"],
    "Layer 2\n(Hidden layer)": ["h2_1", "h2_2", "h2_3"],
    "Layer 3\n(Output layer)": ["y1", "y2"]}

G = nx.DiGraph()

# add nodes
for layer_name, nodes in layers.items():
    for n in nodes:
        G.add_node(n, layer=layer_name)

# add edges
layer_names = list(layers.keys())
for i in range(len(layer_names)-1):
    for a in layers[layer_names[i]]:
        for b in layers[layer_names[i+1]]:
            G.add_edge(a, b)

# position
pos = {}
x_gap = 3

for i, layer_name in enumerate(layer_names):
    nodes = layers[layer_name]
    y_gap = 2
    total_height = (len(nodes)-1)*y_gap
    for j, n in enumerate(nodes):
        pos[n] = (i*x_gap, total_height/2 - j*y_gap)

plt.figure(figsize=(10, 6))

nx.draw_networkx_edges(
    G, pos,
    arrows=True,
    arrowstyle="-|>",
    arrowsize=25,
    width=1.4,
    alpha=0.8,
    min_source_margin=15,
    min_target_margin=15)

node_colors = []
for n in G.nodes:
    layer = G.nodes[n]['layer']
    if "Input" in layer:
        node_colors.append("white")
    else:
        node_colors.append("black")

nx.draw_networkx_nodes(
    G, pos,
    node_shape='o',
    node_size=900,
    node_color=node_colors,
    edgecolors='black')

labels = {}

nx.draw_networkx_labels(G, pos, labels=labels, font_size=10, font_color="black")

for i, layer_name in enumerate(layer_names):
    x = i*x_gap
    plt.text(x, -3, layer_name, ha='center', fontsize=12, fontweight='bold')

plt.axis("off")
plt.tight_layout()
plt.savefig("outputs/2_2_3_2_NN.png", dpi=300, bbox_inches="tight")
plt.show()
```
