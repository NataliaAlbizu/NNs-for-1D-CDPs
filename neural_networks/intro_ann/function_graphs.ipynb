```python
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
```

```python
Path("outputs").mkdir(parents=True, exist_ok=True)
```

```python
x = np.linspace(-10, 10, 400)

sigmoid = 1 / (1 + np.exp(-x))
relu = np.maximum(0, x)
tanh = np.tanh(x)
def heaviside_manual(x):
    return np.where(x > 0, 1, 0)
y = heaviside_manual(x)

# --- Heaviside ---
x_neg = x[x < 0]
y_neg = y[x < 0]
x_pos = x[x > 0]
y_pos = y[x > 0]

plt.figure()
plt.plot(x_neg, y_neg, color='C0', linewidth=2.5)
plt.plot(x_pos, y_pos, color='C0', linewidth=2.5)
plt.scatter([0], [0], color='C0')   # H(0)=0
plt.title("Heaviside Step Function")
plt.xlabel("x")
plt.ylabel("H(x)")
plt.grid(True)
plt.savefig('outputs/heaviside.png')
plt.show()

# --- Sigmoid ---
plt.figure()
plt.plot(x, sigmoid)
plt.title("Sigmoid")
plt.xlabel("x")
plt.ylabel(r"$\sigma(x)$")
plt.grid(True)
plt.savefig('outputs/sigmoid.png')
plt.show()

# --- ReLU ---
plt.figure()
plt.plot(x, relu)
plt.title("ReLU")
plt.xlabel("x")
plt.ylabel("ReLU(x)")
plt.grid(True)
plt.savefig('outputs/relu.png')
plt.show()

# --- Tanh ---
plt.figure()
plt.plot(x, tanh)
plt.title("tanh")
plt.xlabel("x")
plt.ylabel("tanh(x)")
plt.grid(True)
plt.savefig('outputs/tanh.png')
plt.show()
```
