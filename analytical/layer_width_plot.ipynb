```python
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
```

```python
Path("outputs").mkdir(parents=True, exist_ok=True)
```

```python
def layer_width(eps):
    lw = eps*np.log(1/(eps*(1 - np.exp(-1/eps))))
    return lw
```

```python
X = np.array([10**(-7), 10**(-6), 10**(-5), 10**(-4), 10**(-3), 10**(-2), 10**(-1)])
Y = layer_width(X)
```

```python
plt.loglog(X, Y, 'o--')
plt.title("Layer width", pad=12)
plt.xlabel(r"$\varepsilon$")
plt.grid()
plt.grid(which="minor", color="0.9")
plt.savefig('outputs/layer_width.png')
plt.show()
```
