```python
import numpy as np
import matplotlib.pyplot as plt
import sys
from pathlib import Path
```

```python
PROJECT_ROOT = Path.cwd().parent
sys.path.insert(0, str(PROJECT_ROOT))

from utils.auxil import analytical

Path("outputs").mkdir(parents=True, exist_ok=True)
```
